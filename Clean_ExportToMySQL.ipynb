## Importing Dependencies

```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine
```

# Loading Data & Checking for Missing Values & Duplicate Rows

```python
df = pd.read_csv("Walmart.csv")
df.shape
```

## Checking for missing & duplicate values

```python
print(df.isnull().sum())
```

```python
df.duplicated().sum()
```

## Dealing with duplicates & missing values

```python
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)
```

Since our data has minimal duplicate values & rows < 1%, I decided to simply drop these values instead of imputing them.

# Further checks

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

THe "unit_price" column should be a numerical column, this is caused by the dollar sign infront of the values

```python
df['unit_price'] = df['unit_price'].str.replace('$','')
df['unit_price'] = df['unit_price'].astype(float)
print(f' unit_price is a {df['unit_price'].dtype}')
```

## Creating a "total_price" column

```python
df['total_price'] = df['unit_price'] * df['quantity']
df.head()
```

## Exporting to cleaned data as CSV

```python
df.to_csv('walmart_cleaned.csv', index=False)
```

# Exporting Data to MySQL

```python
engine_mysql = create_engine("mysql+pymysql://root@localhost:3306/walmart_db")

try:
    engine_mysql
    print("Connected successfully to MySQL")
except:
    print("Failed")
```

```python
df.to_sql(name="Walmart", con = engine_mysql, if_exists="append", index=False)
```
